```python
import os
import xml.etree.ElementTree as ET

# مسار مجلد الأنوتيشنات والصور
annotations_path = "/root/.cache/kagglehub/datasets/zaraks/pascal-voc-2007/versions/1/voctrainval_06-nov-2007/VOCdevkit/VOC2007/Annotations"
images_path = "/root/root/.cache/kagglehub/datasets/zaraks/pascal-voc-2007/versions/1/voctrainval_06-nov-2007/VOCdevkit/VOC2007/JPEGImages"
output_path = "/root/.cache/kagglehub/datasets/zaraks/pascal-voc-2007/versions/1/voctrainval_06-nov-2007/yolo_annotations"

# التأكد من وجود المجلد لإخراج الأنوتيشنات
os.makedirs(output_path, exist_ok=True)

# قائمة الفئات (كلاسات) من Pascal VOC 2007
classes = ["aeroplane", "bicycle", "bird", "boat", "bottle", "bus", "car", "cat", "chair", "cow", "dog", "horse", "motorbike", "person", "pottedplant", "sheep", "sofa", "train", "tvmonitor"]

def convert_annotation(xml_file):
    tree = ET.parse(xml_file)
    root = tree.getroot()

    # اسم الصورة بدون الامتداد
    filename = root.find("filename").text.replace(".jpg", "")

    # إنشاء ملف YOLO للأنوتيشن
    yolo_file = os.path.join(output_path, f"{filename}.txt")

    with open(yolo_file, "w") as out_file:
        for obj in root.iter("object"):
            class_name = obj.find("name").text
            if class_name in classes:
                # تحديد الـ class_id بناءً على الفئة
                class_id = classes.index(class_name)

                # حساب الـ bounding box
                bbox = obj.find("bndbox")
                xmin = int(bbox.find("xmin").text)
                ymin = int(bbox.find("ymin").text)
                xmax = int(bbox.find("xmax").text)
                ymax = int(bbox.find("ymax").text)

                # تحويل الـ bounding box إلى تنسيق YOLO (normalized)
                img_width = int(root.find("size").find("width").text)
                img_height = int(root.find("size").find("height").text)

                # حساب المركز والابعاد بشكل normalized
                x_center = (xmin + xmax) / 2 / img_width
                y_center = (ymin + ymax) / 2 / img_height
                width = (xmax - xmin) / img_width
                height = (ymax - ymin) / img_height

                # كتابة في الملف
                out_file.write(f"{class_id} {x_center} {y_center} {width} {height}\n")

# تحويل كل الأنوتيشنات
for xml_file in os.listdir(annotations_path):
    if xml_file.endswith(".xml"):
        convert_annotation(os.path.join(annotations_path, xml_file))
```

```python
# تحديد مسار ملف yolov4.cfg
cfg_file_path = '/content/darknet/cfg/yolov4.cfg'

# فتح الملف للتعديل
with open(cfg_file_path, 'r') as file:
    lines = file.readlines()

# تعديل الفلاتر في كل قسم [convolutional] (تصحيح قيم filters)
for i, line in enumerate(lines):
    if line.strip().startswith('filters'):
        lines[i] = 'filters=75\n'  # تصحيح الفلاتر إلى 75 (لـ 20 كلاس + 5)

# تعديل الكلاسات في كل قسم [yolo]
for i, line in enumerate(lines):
    if line.strip().startswith('classes'):
        lines[i] = 'classes=20\n'  # عدد الكلاسات هو 20 في PASCAL VOC

# حفظ التعديلات في ملف cfg
with open(cfg_file_path, 'w') as file:
    file.writelines(lines)

print(f'تم تعديل ملف {cfg_file_path} بنجاح!')
```

```python
# مسار ملف التكوين
cfg_file = 'cfg/yolov4.cfg'

# تعديل القيم المطلوبة
def modify_yolo_config(cfg_file):
    with open(cfg_file, 'r') as f:
        lines = f.readlines()

    # تعديل قيم الـ batch و subdivisions و width و height
    for i, line in enumerate(lines):
        if line.startswith('batch'):
            lines[i] = 'batch=32\n'  # تقليل حجم الـ batch
        elif line.startswith('subdivisions'):
            lines[i] = 'subdivisions=64\n'  # زيادة subdivisions
        elif line.startswith('width'):
            lines[i] = 'width=416\n'  # تقليل العرض
        elif line.startswith('height'):
            lines[i] = 'height=416\n'  # تقليل الارتفاع

    # حفظ التعديلات في الملف
    with open(cfg_file, 'w') as f:
        f.writelines(lines)

    print(f"{cfg_file} تم تعديله بنجاح!")

# تنفيذ التعديل
modify_yolo_config(cfg_file)
```
